```python
import numpy as np
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**3)
```

```python
print(np.sqrt(a)) #to find the sq root
```

```python
arr=np.array([10,20,20,30,40,50]) #defining 2d array
print(np.mean(arr))
print(np.median(arr))
```

```python
print(np.max(arr))
print (np.min(arr))
print(np.sum(arr))
print(np.std(arr)) #standard deviation
print(np.var(arr)) #variance
```
